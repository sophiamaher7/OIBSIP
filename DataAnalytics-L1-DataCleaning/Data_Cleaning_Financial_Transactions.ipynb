```python
text_columns = [
    "Transaction_ID",
    "Customer_ID",
    "Product_Name",
    "Payment_Method",
    "Transaction_Status"]

for column in text_columns:
    df_clean[column] = (
        df_clean[column]
        .astype("string")
        .str.strip())
    
```

```python
valid_products = [
    "Tablet",
    "Smartphone",
    "Laptop",
    "Headphones",
    "Coffee Machine"]

def standardize_product(value):
    if pd.isna(value):
        return value

    value = value.strip().title()

    matches = [
        product for product in valid_products
        if product.lower().startswith(value.lower())
    ]

    if len(matches) == 1:
        return matches[0]

    return value

df_clean["Product_Name"] = df_clean["Product_Name"].apply(
    standardize_product)
```

```python
payment_mapping = {
    "credit card": "Credit Card",
    "creditcard": "Credit Card",
    "pay pal": "PayPal",
    "paypal": "PayPal",
    "cash": "Cash"}

df_clean["Payment_Method"] = (
    df_clean["Payment_Method"]
    .str.lower()
    .map(payment_mapping))
```

## Missing and Invalid Data Handling

Missing and invalid values were handled according to the meaning of each
column. Values that could not be reliably inferred were removed, while
numeric and categorical values were imputed where an appropriate replacement
could be determined.

```python
df_clean["Price"]= df_clean["Price"].astype(str).str.replace("$" , "" , regex=False).str.replace("," , "" , regex=False)
df_clean["Price"]= pd.to_numeric(df_clean["Price"] , errors="coerce")
df_clean["Price"] = df_clean["Price"].round(2)
```

```python
df_clean["Transaction_Date"]= pd.to_datetime(df_clean["Transaction_Date"], errors="coerce")
```

## Missing Data Handling

The missing values were handled according to the meaning and data type of each column.

- **Transaction_ID:** Rows with missing transaction IDs will be removed because the ID is the unique identifier of a transaction and cannot be reliably inferred.

- **Transaction_Date:** Missing and invalid transaction dates will be converted to missing datetime values (`NaT`) and replaced with the median valid transaction date. The transactions will be retained because the date cannot be reliably reconstructed from the available data, but the remaining transaction information is still useful for analysis.

- **Customer_ID:** Missing customer IDs will be replaced with `"Unknown"` because removing these transactions would result in unnecessary data loss.

- **Quantity:** Missing quantities will be replaced with the median quantity because it is a numerical variable and the median is less affected by extreme values.

- **Price:** Price is associated with the product being sold, so missing prices will be imputed using the median price within each product. This preserves differences in typical prices across products. The overall median will be used only as a fallback where a product has no valid price observations.

- **Transaction_Status:** Missing statuses will be replaced with the mode because this is a categorical variable and the most frequent status provides a reasonable replacement.

- No imputation is required for **Product_Name** or **Payment_Method** because they do not contain missing values.

```python
df_clean= df_clean.dropna(subset=["Transaction_ID"])
```

### Transaction Date Handling

The Transaction_Date column contained missing values and invalid dates such as
2023-13-01 and 2025-02-30. Invalid dates were converted to NaT and then
missing/invalid dates were replaced with the median valid transaction date.
Rows were retained because the transaction records still contain useful
information even when the date is unavailable or invalid.

```python
date_median = df_clean["Transaction_Date"].median()

df_clean["Transaction_Date"] = df_clean["Transaction_Date"].fillna(
    date_median)
```

```python
df_clean["Customer_ID"]= df_clean["Customer_ID"].fillna("Unknown")
```

```python
Quantity_Median= df_clean["Quantity"].median()
df_clean["Quantity"]= df_clean["Quantity"].fillna(Quantity_Median)
```

```python
Status_Mode= df_clean["Transaction_Status"].mode()[0]
df_clean["Transaction_Status"]= df_clean["Transaction_Status"].fillna(Status_Mode)
```

```python
Product_Median = df_clean.groupby("Product_Name")["Price"].transform("median")
df_clean["Price"]= df_clean["Price"].fillna(Product_Median)

df_clean["Price"] = df_clean["Price"].fillna(df_clean["Price"].median())
```

```python
df_clean = df_clean.reset_index(drop=True)
```

## Outlier Detection

The IQR method was used to identify statistical outliers in the numeric
columns. Statistical outliers were not automatically treated as errors,
because unusually large transactions may represent legitimate observations.

```python
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)]

    return outliers, lower_bound, upper_bound
```

```python
quantity_outliers, quantity_lower, quantity_upper = (
    detect_outliers_iqr(df_clean, "Quantity"))

price_outliers, price_lower, price_upper = (
    detect_outliers_iqr(df_clean, "Price"))
print(f"Quantity outliers: {len(quantity_outliers)}")
print(f"Price outliers: {len(price_outliers)}")
```

### Value Range Anomalies and Outliers

Negative quantities and negative prices were identified during the range check.
Although these values are unusual for standard sales transactions, they were not
automatically classified as errors because they may represent legitimate
business cases such as returns, refunds, reversals, or transaction adjustments.
The dataset does not provide enough information to reliably determine their
meaning, so these values will be retained and documented as potential anomalies.

For statistical outliers, the IQR method will be used on the numeric columns.
Values identified as statistical outliers will also be retained because an
unusually large or small transaction may still represent a legitimate
transaction. No observations will be removed solely because they are
statistical outliers.

```python
df_clean["Transaction_ID"] = df_clean["Transaction_ID"].astype("string")

df_clean["Transaction_Date"] = pd.to_datetime(
    df_clean["Transaction_Date"])

df_clean["Customer_ID"] = df_clean["Customer_ID"].astype("string")

df_clean["Product_Name"] = df_clean["Product_Name"].astype("string")

df_clean["Quantity"] = df_clean["Quantity"].astype("int64")

df_clean["Price"] = df_clean["Price"].astype("float64")

df_clean["Payment_Method"] = df_clean["Payment_Method"].astype("string")

df_clean["Transaction_Status"] = df_clean["Transaction_Status"].astype("string")
```
